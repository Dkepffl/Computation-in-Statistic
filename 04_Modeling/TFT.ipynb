```python
# 2. 날짜 처리 & time_idx 생성 (필수!)
# 데이터에 'Year', 'Month' 컬럼이 있다고 가정합니다.
# 1996년 1월을 0으로 시작해서 매달 1씩 증가하는 정수 인덱스를 만듭니다.
min_year = data['Year'].min()
data['time_idx'] = (data['Year'] - min_year) * 12 + (data['Month'] - 1)

# 3. 데이터 타입 정리
# ticker 같은 그룹 ID는 문자열(String)이어야 합니다.
data['ticker'] = data['ticker'].astype(str)
data['Year'] = data['Year'].astype(str) # 범주형으로 쓸 거면 str
data['Month'] = data['Month'].astype(str)

# 4. 결측치 처리 (TFT는 NaN을 허용하지 않습니다)
# 간단하게 0이나 평균으로 채웁니다.
data = data.fillna(0)

print("Time Index created. Max time_idx:", data['time_idx'].max())
print(data.head())
```

## **TimeSeriesDataSet 생성**

```python
from pytorch_forecasting import TimeSeriesDataSet, GroupNormalizer

# 설정 변수
MAX_ENCODER_LENGTH = 12  # 과거 12개월(1년) 데이터를 보고
MAX_PREDICTION_LENGTH = 1 # 미래 1개월을 예측

# 피처 리스트 정리 (사용자 데이터 컬럼에 맞춰 자동화)
# 'ret', 'ticker', 'date', 'time_idx', 'Year', 'Month' 등을 제외한 나머지 모든 숫자 변수들
exclude_cols = ['ret', 'ticker', 'date', 'time_idx', 'Year', 'Month', 'permno', 'gvkey']
feature_cols = [c for c in data.columns if c not in exclude_cols]

# 학습 데이터셋 정의 (2021년 이전 데이터만 사용 예시)
training_cutoff = data["time_idx"].max() - MAX_PREDICTION_LENGTH # 마지막 1달은 검증용

training = TimeSeriesDataSet(
    data[data["time_idx"] <= training_cutoff],
    time_idx="time_idx",
    target="ret",
    group_ids=["ticker"], # 각 종목(ticker)별로 시계열이 따로 존재함을 명시

    min_encoder_length=MAX_ENCODER_LENGTH // 2, # 최소 6개월은 데이터가 있어야 학습
    max_encoder_length=MAX_ENCODER_LENGTH,
    min_prediction_length=1,
    max_prediction_length=MAX_PREDICTION_LENGTH,

    # 1. 정적 변수 (시간이 지나도 안 변하는 것)
    static_categoricals=["ticker"],

    # 2. 미래에도 알 수 있는 변수 (시간, 날짜 등)
    time_varying_known_categoricals=["Year", "Month"],
    time_varying_known_reals=["time_idx"],

    # 3. 과거만 아는 변수 (실제 피처들 + 타겟)
    # 미래 시점의 PBR, PER은 모르므로 여기에 넣어야 함
    time_varying_unknown_reals=["ret"] + feature_cols,

    # 타겟 정규화 (종목별로 주가/수익률 스케일이 다를 때 유용)
    target_normalizer=GroupNormalizer(
        groups=["ticker"], transformation="softplus"
    ),

    add_relative_time_idx=True,
    add_target_scales=True,
    add_encoder_length=True,
)

# 검증용 데이터셋 생성 (Training 설정을 그대로 복사하되 데이터만 바꿈)
validation = TimeSeriesDataSet.from_dataset(training, data, predict=True, stop_randomization=True)

# 데이터로더 생성 (배치 사이즈는 GPU 메모리에 맞춰 조절)
batch_size = 64
train_dataloader = training.to_dataloader(train=True, batch_size=batch_size, num_workers=0)
val_dataloader = validation.to_dataloader(train=False, batch_size=batch_size * 10, num_workers=0)

print("Dataset & Dataloader Created!")
```

## **모델 정의 및 학습**

```python
import pytorch_lightning as pl
from pytorch_forecasting.models import TemporalFusionTransformer
from pytorch_forecasting.metrics import RMSE, MAE
from pytorch_lightning.callbacks import EarlyStopping, LearningRateMonitor

# 1. 모델 설정
tft = TemporalFusionTransformer.from_dataset(
    training,
    learning_rate=0.03,
    hidden_size=16,           # 모델 크기 (크면 성능 좋지만 느리고 메모리 많이 먹음)
    attention_head_size=1,
    dropout=0.1,
    hidden_continuous_size=8,
    output_size=1,            # Point Prediction (단일 값 예측)인 경우 1 (Quantile이면 7)
    loss=RMSE(),              # 손실함수
    log_interval=10,
    reduce_on_plateau_patience=4,
)

print(f"Number of parameters in network: {tft.size()/1e3:.1f}k")

# 2. 트레이너 설정
early_stop_callback = EarlyStopping(monitor="val_loss", min_delta=1e-4, patience=10, verbose=False, mode="min")
lr_logger = LearningRateMonitor()

trainer = pl.Trainer(
    max_epochs=30,            # 최대 학습 횟수
    accelerator="auto",       # GPU 있으면 자동 사용
    enable_model_summary=True,
    gradient_clip_val=0.1,
    callbacks=[lr_logger, early_stop_callback],
)

# 3. 학습 시작
print("Starting Training...")
trainer.fit(
    tft,
    train_dataloaders=train_dataloader,
    val_dataloaders=val_dataloader,
)
```

```python
import os
import pandas as pd
import numpy as np
import torch

import lightning.pytorch as pl
from lightning.pytorch.callbacks import EarlyStopping, LearningRateMonitor

from pytorch_forecasting import TimeSeriesDataSet, TemporalFusionTransformer, GroupNormalizer
from pytorch_forecasting.metrics import RMSE
from pytorch_forecasting.data.encoders import NaNLabelEncoder
from sklearn.preprocessing import StandardScaler

# 경고 무시
import warnings
warnings.filterwarnings("ignore")
```

```python
def run_tft_pipeline(year, batch_size=64, max_epochs=20):
    print(f"\n{'#'*20} Start TFT Pipeline: {year} {'#'*20}")

    # 1. 데이터 통합
    train = globals()[f'train{year}']
    valid = globals()[f'valid{year}']
    test  = globals()[f'test{year}']

    train['split'] = 'train'
    valid['split'] = 'valid'
    test['split']  = 'test'

    data = pd.concat([train, valid, test], ignore_index=True)

    # 2. 데이터 클리닝
    data['ret'] = pd.to_numeric(data['ret'], errors='coerce')
    data = data.replace([np.inf, -np.inf], 0.0).fillna(0.0)

    # 타겟 스케일링
    scaler = StandardScaler()
    data['ret_scaled'] = scaler.fit_transform(data[['ret']])
    ret_mean = scaler.mean_[0]
    ret_scale = scaler.scale_[0]
    print(f"Global Scaling Applied. Mean: {ret_mean:.4f}, Std: {ret_scale:.4f}")

    # 3. 전처리
    if 'date' in data.columns:
        data['date'] = pd.to_datetime(data['date'])

    min_date = data['date'].min()
    data['time_idx'] = (data['date'].dt.year - min_date.year) * 12 + (data['date'].dt.month - min_date.month)
    data['time_idx'] = data['time_idx'].astype(int)

    # 짧은 종목 제거
    min_len_req = 13
    cnt_per_ticker = data.groupby('ticker')['time_idx'].count()
    valid_tickers = cnt_per_ticker[cnt_per_ticker >= min_len_req].index
    data = data[data['ticker'].isin(valid_tickers)].reset_index(drop=True)

    # 범주형 변수 처리
    cat_cols = ['ticker', 'Year', 'Month', 'sector']
    exist_cat_cols = [c for c in cat_cols if c in data.columns]

    for c in exist_cat_cols:
        data[c] = data[c].astype(str).astype('category')

    # 4. TimeSeriesDataSet 생성
    train_cutoff = data[data['split']=='train']['time_idx'].max()

    exclude = ['ret', 'ret_scaled', 'date', 'time_idx', 'split', 'permno', 'gvkey'] + cat_cols
    feature_cols = [c for c in data.columns if c not in exclude and pd.api.types.is_numeric_dtype(data[c])]

    max_encoder_length = 12
    max_prediction_length = 1

    categorical_encoders = {
        col: NaNLabelEncoder(add_nan=True) for col in exist_cat_cols
    }

    training = TimeSeriesDataSet(
        data[data['time_idx'] <= train_cutoff],
        time_idx="time_idx",
        target="ret_scaled",
        group_ids=["ticker"],
        min_encoder_length=max_encoder_length // 2,
        max_encoder_length=max_encoder_length,
        min_prediction_length=1,
        max_prediction_length=max_prediction_length,
        static_categoricals=["ticker"],
        time_varying_known_categoricals=[c for c in ["Year", "Month"] if c in data.columns],
        time_varying_known_reals=["time_idx"],
        time_varying_unknown_reals=["ret_scaled"] + feature_cols,

        allow_missing_timesteps=True,
        target_normalizer=None,
        categorical_encoders=categorical_encoders,

        add_relative_time_idx=True,
        add_target_scales=True,
        add_encoder_length=True,
    )

    validation = TimeSeriesDataSet.from_dataset(training, data[data['split'] != 'test'], predict=True, stop_randomization=True)

    train_dataloader = training.to_dataloader(train=True, batch_size=batch_size, num_workers=0)
    val_dataloader = validation.to_dataloader(train=False, batch_size=batch_size*10, num_workers=0)

    # 5. 모델 학습
    tft = TemporalFusionTransformer.from_dataset(
        training,
        learning_rate=0.03,
        hidden_size=8,
        attention_head_size=1,
        dropout=0.1,
        hidden_continuous_size=4,
        output_size=1,
        loss=RMSE(),
        log_interval=10,
        reduce_on_plateau_patience=4,
    )

    # [수정] Callbacks import 경로 변경됨
    early_stop_callback = EarlyStopping(monitor="val_loss", min_delta=1e-4, patience=5, verbose=False, mode="min")
    lr_logger = LearningRateMonitor()

    trainer = pl.Trainer(
        max_epochs=max_epochs,
        accelerator="auto",
        enable_model_summary=True,
        gradient_clip_val=0.1,
        callbacks=[lr_logger, early_stop_callback],
        logger=True
    )

    trainer.fit(tft, train_dataloaders=train_dataloader, val_dataloaders=val_dataloader)

    # 6. 예측 및 결과 저장
    best_model_path = trainer.checkpoint_callback.best_model_path
    best_tft = TemporalFusionTransformer.load_from_checkpoint(best_model_path, weights_only=False)

    # Test 데이터 추출
    test_start_idx = data[data['split']=='test']['time_idx'].min()
    test_subset = data[data['time_idx'] >= test_start_idx - max_encoder_length]
    test_subset = test_subset[test_subset['ticker'].isin(valid_tickers)]

    test_ds = TimeSeriesDataSet.from_dataset(training, test_subset, predict=False, stop_randomization=True)
    test_dataloader = test_ds.to_dataloader(train=False, batch_size=batch_size*10, num_workers=0)

    # 예측
    raw_predictions = best_tft.predict(test_dataloader, mode="prediction")
    pred_vals_scaled = raw_predictions.flatten().cpu().numpy()

    # 역변환
    pred_vals_original = pred_vals_scaled * ret_scale + ret_mean

    # 결과 매칭
    target_rows = test_subset[test_subset['time_idx'] >= test_start_idx].copy()
    target_rows = target_rows.sort_values(by=['ticker', 'time_idx'])

    min_len = min(len(pred_vals_original), len(target_rows))

    result_df = target_rows.iloc[:min_len][['ticker', 'date']].copy()
    result_df['pred_ret'] = pred_vals_original[:min_len]
    result_df['true_ret'] = target_rows.iloc[:min_len]['ret'].values

    save_dir = os.path.join(base_path, '00_predData', 'TFT')
    os.makedirs(save_dir, exist_ok=True)
    result_df.to_csv(os.path.join(save_dir, f'result_{year}.csv'), index=False)

    print(f"[{year}] Done! Saved.")
    return result_df
```
