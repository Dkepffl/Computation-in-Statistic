```python
## Import libaries
import pandas as pd
import numpy as np

import os

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
## Setting
base_path = '/content/drive/MyDrive/CS/'
```

## **CatBoost**

### **데이터 불러오기**

```python
## Load the data
file_path = os.path.join(base_path, 'PredData', 'CatBoost')

result_2018_catboost = pd.read_csv(os.path.join(file_path, 'result_2018.csv'))
result_2019_catboost = pd.read_csv(os.path.join(file_path, 'result_2019.csv'))
result_2020_catboost = pd.read_csv(os.path.join(file_path, 'result_2020.csv'))
result_2021_catboost = pd.read_csv(os.path.join(file_path, 'result_2021.csv'))
```

```python
result_2018_catboost.head()
```

### **R-squared 값 구하기**

```python
from sklearn.metrics import r2_score

y_true_2018 = result_2018_catboost['true_ret']
y_pred_2018 = result_2018_catboost['pred_ret']
r2_2018 = r2_score(y_true_2018, y_pred_2018)

y_true_2019 = result_2019_catboost['true_ret']
y_pred_2019 = result_2019_catboost['pred_ret']
r2_2019 = r2_score(y_true_2019, y_pred_2019)

y_true_2020 = result_2020_catboost['true_ret']
y_pred_2020 = result_2020_catboost['pred_ret']
r2_2020 = r2_score(y_true_2020, y_pred_2020)

y_true_2021 = result_2021_catboost['true_ret']
y_pred_2021 = result_2021_catboost['pred_ret']
r2_2021 = r2_score(y_true_2021, y_pred_2021)

summary_r2 = pd.DataFrame({'Year': [2018, 2019, 2020, 2021], 'R-squared': [r2_2018, r2_2019, r2_2020, r2_2021]})
summary_r2['model'] = 'CatBoost'
display(summary_r2)
```
