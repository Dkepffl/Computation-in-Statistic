## **데이터 필터링**


```python
raw['year'] = raw['date'].dt.year
unique_tickers_per_year_raw = raw.groupby('year')['ticker'].nunique().reset_index()
unique_tickers_per_year_raw.columns = ['Year', 'Number of Unique Tickers']

display(unique_tickers_per_year_raw.head())
display(unique_tickers_per_year_raw.tail())
```

- 시가 총액을 기준으로 매년 Top 1000개의 기업 리스트를 추출

```python
def filter_top_n_yearly(df, num_top=1000):
    filtered_frames = []

    if 'year' not in df.columns:
        df['year'] = pd.to_datetime(df['date']).dt.year

    unique_years = sorted(df['year'].unique())

    for year in unique_years:
        df_year = df[df['year'] == year].copy()

        yearly_me = df_year.groupby('ticker')['me'].median()

        top_tickers = yearly_me.nlargest(num_top).index.tolist()

        # 선정된 기업에 해당하는 데이터만 남김 (그 해 데이터만!)
        df_year_filtered = df_year[df_year['ticker'].isin(top_tickers)]
        filtered_frames.append(df_year_filtered)

    final_df = pd.concat(filtered_frames, axis=0)

    return final_df
```

```python
firm_top1000 = filter_top_n_yearly(raw, num_top=1000) # Top 1000
firm_top1500 = filter_top_n_yearly(raw, num_top=1500) # Top 1500
```

```python
print("\n=== 결과 확인 ===")
print(f"원본 데이터 크기: {raw.shape}")
print(f"Top 1000 필터링 후 크기: {firm_top1000.shape}")
print(f"Top 1500 필터링 후 크기: {firm_top1500.shape}")
```

## **데이터 전처리**

### **결측치/이상치 처리**

```python
## Top 1000
numeric_cols1000 = firm_top1000.select_dtypes(include=np.number)

# inf and -inf
inf_count = numeric_cols1000.apply(lambda x: np.isinf(x).sum())
print("Count of inf / -inf values per numeric column in firm_top1000:")
print(inf_count[inf_count > 0])

# NaN
nan_count = firm_top1000.isnull().sum()
print("\nCount of NaN values per column in firm_top1000:")
print(nan_count[nan_count > 0])
```

```python
## Top 1500
numeric_cols1500 = firm_top1500.select_dtypes(include=np.number)

# inf and -inf
inf_count = numeric_cols1500.apply(lambda x: np.isinf(x).sum())
print("Count of inf / -inf values per numeric column in firm_top1500:")
print(inf_count[inf_count > 0])

# NaN
nan_count = firm_top1500.isnull().sum()
print("\nCount of NaN values per column in firm_top1500:")
print(nan_count[nan_count > 0])
```

```python
## inf, -inf 값을 결측치로 처리
firm_top1000=firm_top1000.replace([np.inf, -np.inf], np.nan)
firm_top1500=firm_top1500.replace([np.inf, -np.inf], np.nan)
```

```python
## Forward Fill : 이전 값으로 채우기
firm_top1000[numeric_cols1000.columns] = firm_top1000.groupby('permno')[numeric_cols1000.columns].ffill()
firm_top1500[numeric_cols1500.columns] = firm_top1500.groupby('permno')[numeric_cols1500.columns].ffill()

## 그래도 남은 값은 0으로
firm_top1000[numeric_cols1000.columns] = firm_top1000[numeric_cols1000.columns].fillna(0)
firm_top1500[numeric_cols1500.columns] = firm_top1500[numeric_cols1500.columns].fillna(0)
```

## **데이터 저장**

```python
# year 변수 제거
firm_top1000.drop('year', axis=1, inplace=True)
firm_top1500.drop('year', axis=1, inplace=True)
```
