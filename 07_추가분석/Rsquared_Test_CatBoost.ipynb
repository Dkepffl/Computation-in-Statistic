```python
## Import libaries
import pandas as pd
import numpy as np

import os

import matplotlib.pyplot as plt
import seaborn as sns
```

## **CatBoost 01**
- Macro HS : 50-20
- Firm HS : 20-20

### **데이터 불러오기**

```python
## Load the data
file_path = os.path.join(base_path, 'PredData', '01_CatBoost_M5020_F2020')

result_2018_catboost01 = pd.read_csv(os.path.join(file_path, 'result_2018.csv'))
result_2019_catboost01 = pd.read_csv(os.path.join(file_path, 'result_2019.csv'))
result_2020_catboost01 = pd.read_csv(os.path.join(file_path, 'result_2020.csv'))
result_2021_catboost01 = pd.read_csv(os.path.join(file_path, 'result_2021.csv'))
```

```python
result_2018_catboost01.head()
```

### **R-squared 값 구하기**

```python
from sklearn.metrics import r2_score

y_true_2018_01 = result_2018_catboost01['true_ret']
y_pred_2018_01 = result_2018_catboost01['pred_ret']
r2_2018_01 = r2_score(y_true_2018_01, y_pred_2018_01)

y_true_2019_01 = result_2019_catboost01['true_ret']
y_pred_2019_01 = result_2019_catboost01['pred_ret']
r2_2019_01 = r2_score(y_true_2019_01, y_pred_2019_01)

y_true_2020_01 = result_2020_catboost01['true_ret']
y_pred_2020_01 = result_2020_catboost01['pred_ret']
r2_2020_01 = r2_score(y_true_2020_01, y_pred_2020_01)

y_true_2021_01 = result_2021_catboost01['true_ret']
y_pred_2021_01 = result_2021_catboost01['pred_ret']
r2_2021_01 = r2_score(y_true_2021_01, y_pred_2021_01)

summary_r2_01 = pd.DataFrame({'Year': [2018, 2019, 2020, 2021], 'R-squared': [r2_2018_01, r2_2019_01, r2_2020_01, r2_2021_01]})
summary_r2_01['HS'] = 'M5020_F2020'
display(summary_r2_01)
```
