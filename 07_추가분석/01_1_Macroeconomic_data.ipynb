### **시차(lag) 반영**
- 다음해의 1월에 당해 12월이 채워져 있지 않았다.
- lag 1 적용

```python
columns_to_lag1 = [col for col in common_columns_list if col not in ['sasdate']]
len(columns_to_lag1)
```

```python
def lag_df(df, columns_to_lag, n):
  for col in columns_to_lag :
    df[col] =df[col].shift(n)
  df = df[['sasdate'] + columns_to_lag]
  df = df[(df['sasdate'] >= '1996-07-01')]

  return(df)
```

```python
df2018_final = lag_df(df2018_filtered, columns_to_lag1, n=1)
df2019_final = lag_df(df2019_filtered, columns_to_lag1, n=1)
df2020_final = lag_df(df2020_filtered, columns_to_lag1, n=1)
df2021_final = lag_df(df2021_filtered, columns_to_lag1, n=1)
```
